```sage
# Declare variables and assumptions
var('r p mu')
assume(r > 0)
assume(p > 0, p < 1)
assume(mu > 0, mu < 1)

# Define the function
f = mu*log(p + (1-p)*r) + (1-mu)*log((1-p)/r + p)
df = diff(f, r)
show(df)
sol = simplify(solve(df == 0 , r))
show(sol)
fsub = f.subs(sol[1]).simplify()
fsub05 = fsub.subs(p==1/2).simplify()
show(fsub05)
fsub01 = fsub.subs(p==0.1).simplify()
a5 = plot(-fsub05, 0.01, 0.99,  legend_label='$p$ = 0.5',color='green')
a1 = plot(-fsub01, 0.01, 0.99,  legend_label='$p$ = 0.1')
a75= plot(-fsub.subs(p==0.75), 0.01, 0.99, legend_label='$p$ = 0.75', color='orange')
(a5+a1 + a75).show(
            legend_loc='upper center',
            legend_font_size=10,
            title='$I(\mu_1)$ for different $p$',
            axes_labels=['$\mu_1$', '$I(\mu_1)$']
            )
```
