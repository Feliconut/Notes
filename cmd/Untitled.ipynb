```python
var('e1, e2')
function('F')(e1, e2)

e_ = lambda g: e1 * diff(g, e2)
f_ = lambda g: e2 * diff(g, e1)
h_ = lambda g: e1 * diff(g, e1) + e2 * diff(g, e2)

C_ = lambda g: e_(f_(g)) + f_(e_(g)) + h_(h_(g))/2

C_(F(e1,e2)).expand()
```
