```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
p = 0.3
mu = 2.0
mu_norm = 0.0
sigma_norm = 1.0

# Domain for plotting
z_vals = np.linspace(0.001, 0.999, 200)  # for Bernoulli/binomial normalized
z_pois = np.linspace(0.001, 6, 200)      # for Poisson
z_norm = np.linspace(-3, 3, 200)         # for Normal

# Rate functions
I_bern = z_vals * np.log(z_vals/p) + (1-z_vals)*np.log((1-z_vals)/(1-p))

I_pois = z_pois * np.log(z_pois/mu) - z_pois + mu

I_norm = (z_norm - mu_norm)**2 / (2*sigma_norm**2)

# Plot
plt.figure(figsize=(14,4))

plt.subplot(1,3,1)
plt.plot(z_vals, I_bern, label="Bernoulli rate")
plt.axvline(p, color="r", linestyle="--", label="mean p")
plt.xlabel("x")
plt.ylabel("I(x)")
plt.title("Bernoulli(p=0.3)")
plt.legend()

plt.subplot(1,3,2)
plt.plot(z_pois, I_pois, label="Poisson rate")
plt.axvline(mu, color="r", linestyle="--", label="mean μ")
plt.xlabel("z")
plt.ylabel("I(z)")
plt.title("Poisson(μ=2)")
plt.legend()

plt.subplot(1,3,3)
plt.plot(z_norm, I_norm, label="Normal rate")
plt.axvline(mu_norm, color="r", linestyle="--", label="mean μ")
plt.xlabel("z")
plt.ylabel("I(z)")
plt.title("Normal(μ=0, σ²=1)")
plt.legend()

plt.tight_layout()
# plt.show()

plt.savefig("src/rate_functions.png")
```
